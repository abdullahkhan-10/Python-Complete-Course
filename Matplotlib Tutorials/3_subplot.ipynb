# Matplotlib Subplot

To draw mutiples plots in one figure we use subplot() function.

The subplot() function takes three arguments that describes the layout of the figure.

The layout is organized in rows and columns, which are represented by the first and second argument. The third argument represents the index of the current plot.

```python
import matplotlib.pyplot as plt

# Product A sales
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

#the figure has 1 row, 2 columns, and this plot is the first plot.
plt.subplot(1,2,1)
plt.plot(months, sales)
plt.title("Figure 1")

# Product B sales
months = [1,2,3,4,5]
sales = [100, 250, 150, 200, 300]

#the figure has 1 row, 2 columns, and this plot is the second plot.
plt.subplot(1,2,2)
plt.plot(months, sales)
plt.title("Figure 2")

plt.show()
```

## Top of ecah other

if we want a figure with 2 rows an 1 column (meaning that the two plots will be displayed on top of each other instead of side-by-side), we can write the syntax like this:

```python
# Product A sales
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

#the figure has 2 row, 1 columns, and this plot is the first plot.
plt.subplot(2, 1, 1)
plt.plot(months, sales)
plt.title("Figure 1")

# Product B sales
months = [1,2,3,4,5]
sales = [100, 250, 150, 200, 300]

#the figure has 2 row, 1 columns, and this plot is the second plot.
plt.subplot(2, 1, 2)
plt.plot(months, sales)
plt.title("Figure 2")

plt.show()
```

## Super Title

We can add a title to the entire figure with the suptitle() function:

```python
# Product A sales
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.subplot(1,2,1)
plt.plot(months, sales)
plt.title("Figure 1")

# Product B sales
months = [1,2,3,4,5]
sales = [100, 250, 150, 200, 300]

plt.subplot(1,2,2)
plt.plot(months, sales)
plt.title("Figure 2")

plt.suptitle("My Company Sales")
plt.show()
```
