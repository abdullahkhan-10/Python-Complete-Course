# Matplotlib Pie Charts

With Pyplot, We can use the pie() function to draw pie charts:

```python
import matplotlib.pyplot as plt

# months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.pie(sales)
plt.show()
```

## Label 

The labels parameter must be an array with one label for each wedge:

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.pie(sales, labels = months)
plt.show()
```

## Wedget percentage

To show the percentage of each wedge we will use the parameter (autopct) inside pie() function.

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.pie(sales, labels = months, autopct = "%1.1f%%")
plt.title("Monthly sale data")
plt.show()
```

## Wedge Color

To display our own choosen color instead of default colors we will use the (colors) parameter inside pie() functions.

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

my_colors = ["skyblue", "gray", "coral", "lightgreen", "gold"]

plt.pie(sales, labels = months, autopct = "%1.1f%%", colors= my_colors)
plt.show()
```

## Legend 

To add a list of explanation for each wedge, use the legend() function:

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.pie(sales, labels = months, autopct = "%1.1f%%")
plt.legend()
plt.show()
```

```python
# To add a header to the legend, add the title parameter to the legend function.

months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.pie(sales, labels = months, autopct = "%1.1f%%")
plt.legend(title = "Monthly Sales")
plt.show()
```

## Explode Wedge 

If we want one of the wedges to stand out? The explode parameter allows you to do that.

The explode parameter, if specified, and not None, must be an array with one value for each wedge.

Each value represents how far from the center each wedge is displayed:

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

# the first wedge will be stand out from pie chart by 0.2.
my_explode = [0.2, 0, 0, 0, 0]

plt.pie(sales, labels = months, autopct = "%1.1f%%", explode = my_explode)
plt.show()
```
