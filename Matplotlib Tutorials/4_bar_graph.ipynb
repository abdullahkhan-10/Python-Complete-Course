# Matplotlib Bars

With Pyplot, We can use the bar() function to draw bar graphs

Bar graphs are commonly utilized for:

1. Comparing Categories: Showing differences between specific groups (e.g., sales figures for different store branches).
2. Tracking Trends: Monitoring large changes in data over a period of time (e.g., monthly temperature changes).
3. Frequency Distributions: Counting how often a specific category or item occurs.
4. Part-to-Whole Relationships: Using stacked bars to show how subcategories make up a total (e.g., a demographic breakdown of a population)

```python
import matplotlib.pyplot as plt

months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.bar(months, sales)
plt.show()
```

## Horizontal Bars

If we want the bars to be displayed horizontally instead of vertically, use the barh() function:

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.barh(months, sales)
plt.show()
```

## Bar Color

The bar() and barh() take the keyword argument color to set the color of the bars:

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.bar(months, sales, color = "orange")
plt.show()
```

```python
# Color Hex
# We can also use Hexadecimal values to decide the color of bars.

months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.bar(months, sales, color = "#4ABF60")
plt.show()
```

## Bar Width

The bar() takes the keyword argument width to set the width of the bars:

```python
months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.bar(months, sales, width = 0.2)
plt.show()
```

```python
# Note: For horizontal bars, use height instead of width

months = ["M1", "M2", "M3", "M4", "M5"]
sales = [120, 200, 150, 250, 190]

plt.barh(months, sales, height = 0.3)
plt.show()
```
