# Matplotlib Histograms

A histogram is a graph showing frequency distributions. It is a graph showing the number of observations within each given interval.

The hist() function will use an array of numbers to create a histogram, the array is sent into the function as an argument.

Example: Let's suppose we have marks of 20 students, We might end up with a histogram like this:

```python
import matplotlib.pyplot as plt

student_marks = [55, 76, 87, 88, 67, 52, 78, 61, 64, 69, 85, 73, 84, 79, 58, 63, 75, 72, 77, 81]

plt.hist(student_marks)
plt.xlabel("Marks Distribution")
plt.ylabel("Number of students")
plt.title("Marks Distribution of students")

plt.show()
```

## Number of Bins

To disply how much boxes/bins on the histogram we will the bins parameter to hist() function.

```python
import matplotlib.pyplot as plt

student_marks = [55, 76, 87, 88, 67, 52, 78, 61, 64, 69, 74, 73, 84, 79, 78, 63, 85, 72, 77, 81]

plt.hist(student_marks, bins = 5)
plt.xlabel("Marks Distribution")
plt.ylabel("Number of students")
plt.title("Marks Distribution of students")

plt.show()
```

## Color and border


```python
student_marks = [55, 76, 87, 88, 67, 52, 78, 61, 64, 69, 74, 73, 84, 79, 78, 63, 85, 72, 77, 81]

plt.hist(student_marks, bins = 5, color = "coral", edgecolor = "black")
plt.xlabel("Marks Distribution")
plt.ylabel("Number of students")
plt.title("Marks Distribution of students")

plt.show()
```
