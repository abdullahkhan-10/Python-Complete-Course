# Pyplot Function

Matplotlib Pyplot has some functions that are used to convert numbers into graph and charts.

Below are the built in Functions of pyplot

### plot()


```python
# plt.plot(x,y, color= "name", linestyle = "", linewidth = value, marker = "o", label = "")

import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.plot(months, sales, color = "blue", linestyle = "--", linewidth = 2, marker = "o", label = "2026 sales data")
plt.show()
```

### xlabel() and ylabel()

```python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [150, 200, 120, 210, 250]
plt.xlabel("Months")
plt.ylabel("Sales per Month")

plt.plot(months, sales, color = "green", linestyle = "--", linewidth = 2, marker = "o", label = "2026 sales data")
plt.show()
```

### title()

```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]
plt.xlabel("Months")
plt.ylabel("Sales per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales, color = "purple", linestyle = "--", linewidth = 2, marker = "o", label = "2026 sales data")
plt.show()
```
