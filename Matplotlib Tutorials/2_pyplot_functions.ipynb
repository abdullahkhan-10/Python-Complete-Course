# Pyplot Function

Matplotlib Pyplot has some functions that are used to convert numbers into graph and charts.

Below are the built in Functions of pyplot

## plot()

The plot() function is used to draw points (markers) in a diagram. By default, the plot() function draws a line from point to point.

The function takes parameters for specifying points in the diagram.

Parameter 1 is an array containing the points on the x-axis.

Parameter 2 is an array containing the points on the y-axis.

```python
# plt.plot(x,y, color= "name", linestyle = "", linewidth = value, marker = "o", label = "")

import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.plot(months, sales, color = "blue", linestyle = "--", linewidth = 2, marker = "o", label = "2026 sales data")
plt.show()
```

### Plotting Without Line

To plot only the markers, you can use shortcut string notation parameter 'o', which means 'rings'.


```python
import matplotlib.pyplot as plt 

months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.plot(months, sales, "o")
plt.show()
```

## xlabel() and ylabel()

With Pyplot, We can use the xlabel() and ylabel() functions to set a label for the x- and y-axis.

```python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [150, 200, 120, 210, 250]
plt.xlabel("Months")
plt.ylabel("Sales per Month")

plt.plot(months, sales, color = "green", linestyle = "--", linewidth = 2, marker = "o", label = "2026 sales data")
plt.show()
```

## title()

With Pyplot, We can use the title() function to set a title for the plot.

```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sales per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales, color = "purple", linestyle = "--", linewidth = 2, marker = "o", label = "2026 sales data")
plt.show()
```

### Font Properties for Title and Labels

We can use the fontdict parameter in xlabel(), ylabel(), and title() to set font properties for the title and labels.

```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

font1 = {'family':'serif','color':'red','size':20}
font2 = {'family':'serif','color':'blue','size':15}

plt.xlabel("Months", fontdict = font2)
plt.ylabel("Sales per Month", fontdict = font2)
plt.title("Monthly sale data report", fontdict = font1)

plt.plot(months, sales, color = "blue", marker = "o", label = "2026 sales data")
plt.show()
```

### Position the Title

We can use the loc parameter in title() to position the title.


```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sales per Month")
plt.title("Monthly sale data report", loc = "left")

plt.plot(months, sales, color = "blue", marker = "o", label = "2026 sales data")
plt.show()
```

## Grid Lines
With Pyplot, We can use the grid() function to add grid lines to the plot.


```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sale per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales)
plt.grid()
plt.show()
```

### Specify Grid Lines

We can use the axis parameter in the grid() function to specify which grid lines to display.


```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sale per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales)
plt.grid(axis = "x")
plt.show()
```

```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sale per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales)
plt.grid(axis = "y")
plt.show()
```

### Line Properties for the Grid

We can also set line roperties for grid lines like (color, line style and line width).

```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sale per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales, marker = "o")
plt.grid(color = "red", linestyle = "--", linewidth = "0.5")
plt.show()
```

### To limit value on x axis and y axis

```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sale per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales, marker = "o")
plt.grid()
plt.xlim(1,5)
plt.ylim(0,300)

plt.show()
```

### Replace value on x axis and y axis

```python
months = [1,2,3,4,5]
sales = [120, 200, 150, 250, 190]

plt.xlabel("Months")
plt.ylabel("Sale per Month")
plt.title("Monthly sale data report")

plt.plot(months, sales, marker = "o")
plt.grid()
plt.xticks([1,2,3,4,5], ["M1", "M2", "M3", "M4", "M5"])

plt.show()
```
