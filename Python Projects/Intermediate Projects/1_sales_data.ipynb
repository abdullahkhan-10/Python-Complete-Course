```python
import pandas as pd

sales_data = pd.read_excel("1_Sales_Data.xlsx")

sales_data.head()
```

```python

sales_data["Manager"] = sales_data["Manager"].str.strip().str.replace(r"\s+", " ", regex= True)


sales_data.drop(0, inplace=True)


sales_data.head()
```

```python
sales_data["Quantity"] = sales_data["Quantity"].astype(float)

sales_data["Quantity"] = sales_data["Quantity"].round()

sales_data["Quantity"] = sales_data["Quantity"].astype(int)

sales_data["Order ID"] = sales_data["Order ID"].astype(int)

# Convert the datatype of "Date" column from object to datetime.
sales_data["Date"] = pd.to_datetime(sales_data["Date"])

sales_data.info()

# Now the data has been clean, Now we will do analysis 
```

### Q) 2. Disply the most selling product by "Quantity" and by "Revenue".

```python
sales_data.head(2)
```

```python
sales_data.groupby("Product")
```

```python
sales_data.groupby("Product")["Quantity"].sum()
```

```python
most_quantity = sales_data.groupby("Product")["Quantity"].sum()

most_quantity
```

```python
type(most_quantity)
```

```python
# The data is in Series formate, We have to convert it Datafram formate to do visualization.

most_quantity = most_quantity.reset_index()

most_quantity

# Now the data is in Dataframe formate.
```

```python
type(most_quantity)
```

```python
# Now show it on Bar graph

import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))
plt.bar(most_quantity["Product"], most_quantity["Quantity"])
plt.title("Most Selling Products by Quantity")
plt.xlabel("Products")
plt.ylabel("Quantity")
```
