```python
import pandas as pd

sales_data = pd.read_excel("1_Sales_Data.xlsx")

sales_data.head()
```

```python
sales_data.drop(columns= "Unnamed: 0", inplace=True)

sales_data.columns = sales_data.loc[0]

sales_data.head()
```

```python

sales_data["Manager"] = sales_data["Manager"].str.strip().str.replace(r"\s+", " ", regex= True)


sales_data.drop(0, inplace=True)


sales_data.head()
```

```python
sales_data["Quantity"] = sales_data["Quantity"].astype(float)

sales_data["Quantity"] = sales_data["Quantity"].round()

sales_data["Quantity"] = sales_data["Quantity"].astype(int)

sales_data["Order ID"] = sales_data["Order ID"].astype(int)

# Convert the datatype of "Date" column from object to datetime.
sales_data["Date"] = pd.to_datetime(sales_data["Date"])

sales_data.info()

# Now the data has been clean, Now we will do analysis 
```
