# Spotify Data Analysis

The dataset contain all information about spotify, We will analyze and find meaningful insight for decision making.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
spotify_data = pd.read_csv("2_Spotify_Data.csv")

spotify_data
```

## First we will understand the data

```python
spotify_data.shape
```

```python
spotify_data.info()
```

```python
spotify_data.describe()
```

```python
spotify_data.columns
```

```python
# Remove unwanted columns.

spotify_data.drop(columns = ["Unnamed: 0", "Url_spotify", "Url_youtube", "Uri"], inplace= True)
```

```python
spotify_data.head()
```

```python
# Check missing values in each columns.

spotify_data.isnull().sum()
```

```python
# Fill the missing values with 0 in Likes and Comments Columns.

spotify_data["Likes"] = spotify_data["Likes"].fillna(0)

spotify_data["Comments"] = spotify_data["Comments"].fillna(0)
```

```python
spotify_data.isnull().sum()
```

```python
# Now the rest we will remove all the rows with missing values.

spotify_data.dropna(inplace= True)
```

```python
spotify_data.isnull().sum()
```

```python
spotify_data.shape

# As we remove the rows with null/missing values, we have left with 19298 rows out of 20718 rows.
```

## Data Analysis

As we have understand and clean our data, Now we are ready to analyze it and find meaningfull insights.

### Q) 1. Show Top 10 Artists with highest views

```python
spotify_data.head(2)
```

```python
artist_views = spotify_data.groupby("Artist")["Views"].sum()
```

```python
artist_views
```

```python
# Now we have to sort it for Top 10 Artist in term of views.

# First we will sort by descinding order
artist_sorted = artist_views.sort_values(ascending= False)

# Show the Top 10.
artist_sorted.head(10)

# Note. The views are in Billion.
```

### Q) 2. Disply Top 10 Tracks, with highest streams on spotify.

```python
# We will solve this question with different way.
# First we will create a dataframe with 2 columns "Track" and "Stream"

new_dataframe = spotify_data[["Track", "Stream"]]

new_dataframe
```

```python
# Now we will sort the new_dataframe wrt to Stream column.

stream_sorted = new_dataframe.sort_values(by= ["Stream"], ascending= False).head(10)

stream_sorted
```

### Q) 3. What are the most common album types on spotify, How many track being to each album type.

```python
spotify_data.head(2)
```

```python
# First we will check all the unique values in the "Album_type" column.

spotify_data["Album_type"].unique()

# So we have three types of album. 
```

### Q) 7. Which top 6 tracks have the highest Like-to-Views Ratio.

```python
spotify_data.head(2)
```

```python
# First we will create dataframe for the three columns.

lv_track = spotify_data[["Track", "Likes", "Views"]]

lv_track
```

```python
# Now we will create Ratio of Like-to-Views.

lv_track["lv_ratio"] = spotify_data["Likes"]/spotify_data["Views"] * 100
```

```python
lv_track
```

```python
# Show the Top 6 ratios.

lv_track.sort_values(by = "lv_ratio", ascending = False ).head(6)

# Note. lv_ratio 24.9 mean that out of 100 views the track has 24 Likes.
```
