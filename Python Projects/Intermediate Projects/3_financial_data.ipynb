# AI Companies Financial Data

The dataset include the Financial Data of some of the Top Artificial Intelligence Companies Like OpenAI, Google Gemini and Meta LLaMA.

We will analyse the data and Find the Revenue earned, Total R & D Spending, Impact on stock, Expenditure vs Revenue and more..

```python
import pandas as pd
```

```python
financial_data = pd.read_csv("3_Financial_Data.csv")

financial_data
```

```python
financial_data.describe()
```

```python
financial_data.info()
```

```python
# As we have the Date column in object formate.
# First we will convert it to Datetime formate.

financial_data["Date"] = pd.to_datetime(financial_data["Date"])
```

```python
financial_data.info()
```

```python
financial_data.head()
```

```python
# Show how many companies we have in the dataset.

financial_data["Company"].unique()
```

```python
# Create a new column for "Year" only.

financial_data["Year"] = financial_data["Date"].dt.year
```

```python
financial_data.head()
```

## Data Analysis

### Q) 1. Display  when GPT-4 was released.

```python
financial_data[financial_data["Event"] == "GPT-4 release"]
```

### Q) 2. How much the company spend on R & D

```python
financial_data.head()
```

```python
financial_data.groupby("Company")["R&D_Spending_USD_Mn"].sum()

# These are the R&D Spending in millions.
```

```python
# To convert it to Billions we will divide it by 1000. 

rd_spending = financial_data.groupby("Company")["R&D_Spending_USD_Mn"].sum()/1000

rd_spending
```

```python
# Draw Bar plot to show the R&D spending in bar graph.

import matplotlib.pyplot as plt

plt.bar(rd_spending.index, rd_spending.values, color= ["red", "blue", "black"])
plt.title("R&D Spending by the Companies")
plt.xlabel("Company")
plt.ylabel("Amount in USD_$Bn")

plt.show()
```

### Q) 3. Total Revenue earned by each Company.

```python
financial_data.head(2)
```

```python
# Revenue of each company in Billion dollars.

comp_revenue = financial_data.groupby("Company")["AI_Revenue_USD_Mn"].sum()/1000

comp_revenue

# So Google has the most revenue of approximately 284.49 Billion Dollars.
```

```python
# Display it on Bar Graph.

plt.bar(comp_revenue.index, comp_revenue.values, color= ["green", "blue", "gray"], width= 0.5)
plt.title("Revenue earned by company")
plt.xlabel("Company")
plt.ylabel("Revenue in USD_$Bn")


plt.show()
```

```python
# Comparison of Expenditure and Revenue of all companies.

plt.figure(figsize= (10,4))

plt.subplot(1,2,1)
plt.bar(rd_spending.index, rd_spending.values, color= ["red", "blue", "black"])
plt.title("R&D Spending by the Companies")
plt.xlabel("Company")
plt.ylabel("Amount in USD_$Bn")

plt.subplot(1,2,2)
plt.bar(comp_revenue.index, comp_revenue.values, color= ["green", "blue", "gray"], width= 0.5)
plt.title("Revenue earned by company")
plt.xlabel("Company")
plt.ylabel("Revenue in USD_$Bn")


plt.show()
```

### Q) 4. Show the date wise impact on stock.

```python
# show in graph

plt.figure(figsize= (10,4))

plt.plot(financial_data["Date"], financial_data["Stock_Impact_%"])
plt.title("Change in stock value")
plt.xlabel("Date in year")
plt.ylabel("Stock_Impact_%")

plt.show()
```

### Q) 5. Create Separate dataset for each company. and Show their stock impact in %
