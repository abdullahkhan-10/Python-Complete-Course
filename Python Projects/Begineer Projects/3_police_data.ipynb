The dataset obtain data from a police chek post, We will and analyze it and find information for decision making.

### 1. Data Ceaning

Remove the column the column that only contains missing values.

```python
# To check the total number of null values in columns we will use sum() function with isnull() function.

police_data.isnull().sum()
```

```python
# Syntax to remove the colummn with null values. We will use the drop() method.

police_data.drop( columns = "country_name")
```

### 2. Filtering and value counts

We will see for Speeding, Who stopped more often Men or Women?

```python
# Speeding is obtain in "violation" column, so we will consider this column
# Then we will filter the violation colume for the word "Speeding".

police_data[police_data.violation == "Speeding"]
```

```python
# Now we will count how many Men and Women have stopped for Speeding, which are in "driver_gender" colume.

police_data[police_data.violation == "Speeding"].driver_gender.value_counts()
```

### 3. Grouping

Does gender affect who gets searched during a stop.

```python
import pandas as pd
police_data = pd.read_csv("3_Police_Data.csv")

police_data
```

```python
# First we have to check how many time the total search has conducted.

police_data.search_conducted.value_counts()
```

```python
# We will analyze two column "driver_gender" and "search_conducted".capitalize.
# To see search conducted against men and women

police_data.groupby("driver_gender").search_conducted.sum()
```

### 4. Mapping and Data-type casting

Find the mean of the column stop_duration.

```python
police_data.head(2)
```

```python
# To check how many unique value are present in "stop_duration" column.
police_data.stop_duration.value_counts()
```

```python
# We will be dealing with column "stop_duration" which has a data type string, we can not find mean for string.
# So we have to convert it to integer to find the mean.

police_data["stop_duration"] = police_data["stop_duration"].map({"0-15 Min" : 8, "16-30 Min ": 25, "30+ Min": 45})
```

```python
police_data
```

```python
# Now we can easily find the mean of "stop_duration" column.

police_data["stop_duration"].mean()
```
