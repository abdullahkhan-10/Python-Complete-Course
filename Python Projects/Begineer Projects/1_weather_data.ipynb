## Q) 9. Find all instances when the "Snow" was recorded.

```python
# We can solve the problem with three methods.
# 1. value_counts()

data["Weather Condition"].value_counts()

# It will show that the snow is recorded 390 times.
```

```python
# 2. Filtering

data[data["Weather Condition"] == "Snow"]
```

```python
# We have seen above that there are some other columns that have snow recorded with rain, showers and fog,
# To record that snow as well we will use the (str.contains) method.

data[data["Weather Condition"].str.contains("Snow")]
```

## Q) 10. Find all instances when "Wind Speed is above 24" and " Visibility is 25"

```python
data.head(2)
```

```python
# We will use the and (&) operator, because we have to find result for two columns.
# We will see only those value where wind speed is above 25 and visibility is 25.

data[(data["Wind Speed_km/h"] > 24) & (data["Visibility_km"] == 25)]
```
