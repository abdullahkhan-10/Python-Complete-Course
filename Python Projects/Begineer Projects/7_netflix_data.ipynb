### Q) 4. In which year the highest number of TV Show and Movies were released,  Show with bar graph.

```python
# As we have space in some values in "Release_Date" column, first we have to remove those spaces. 

netflix_data["New_Date"] = pd.to_datetime(
    netflix_data["Release_Date"].astype(str).str.strip(),
    format="%B %d, %Y",
    errors="coerce"
)
```

```python
# To find the number of all TV Show and Movies Release in each year, we will use value_counts() function.
# value_counts() function is used to counts the occurance of all individual  element in column.

netflix_data["New_Date"].dt.year.value_counts()

# So in 2019 there was a total of 2154 TV Show and Movies released.
```

### Q) 12. What is the maximum duration of Movie/Show on Netflix.

```python
netflix_data.head(2)
```

```python
# We will consider the Duration column
# First we will display all the unique duration from the column.

netflix_data["Duration"].unique()
```

```python
netflix_data.dtypes
```

```python
# As the Duration column is an object data type , we have to convert it to numeric formate.
# we will use str.split() method.
# In str.split(" ") method we have to pass separater which in this case is blanke space.

netflix_data[["Minutes", "Units"]] = netflix_data["Duration"].str.split(" ", expand=True)

# We have created two columns "Minutes" which contain the numeric values and "Units" which contain string values.
```

```python
netflix_data.head(2)
```

```python
# Now we can easily find the maximum duration of Show/Movies.

netflix_data["Minutes"].max()

# As per the available dataset 99 minutes is the maximum duration of shows/Movies.
```
